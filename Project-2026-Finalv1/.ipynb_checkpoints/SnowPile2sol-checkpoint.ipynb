# Heat transfer analysis of the saw dust on top of the pile

A similar analysis as the building is to be performed for the snow pile, which is assumed to be insulated with  wood chip and buried. The conduction through the wood chips determines the melt.

## Assumptions
* The snow pile melts vertically only, meaning that the upper surface remains horizontal throughout the process
* The effect of humidity in the insulation layer is not taken into account by lack of correlation able to capture this critical dynamics. This assumption diminishes the effective insulation property of the wood chips.
* The snow is at $\mathrm{0^\circ C}$ throughout the pile except on the edges, in what is called the mushy layer (mix of water and snow/ice). This effect of this layer on heat transfer is neglected. The temperature is assumed to be uniform
* Assume that the top of the pile, protected by a layer of sawdust (at least $30\mathrm{cm}$) is melting uniformely. 
* Note that the snow pile cannot grow, only melt.
* Assume permafrost (no melting from the ground)

## Melt equation
Based on these assumptions, the rate of melt is computed as the time variation of the volume of the pile assumed to be a parallelogram with right angles and constant horizontal dimensions $L_{x}$ and $L_y$. Using $H$ as the height the melt equation is
$$
\rho_{snow}h_{L,snow}L_xL_y\frac{dH}{dt}=-\max\left[q''_{wood}(t),0\right](L_xL_y)-q_{cool}(t)
$$
where
* $L_x$ and $L_y$ are the horizontal dimensions of the pile
* $h_{L,snow}$ is the latent heat of snow
* $q_{cool}(t)$ is the cooling load required to reduce the fluid temperature

# Write your code to calculate the evolution of height as a function of time.


$$
\frac{dH}{dt}=\frac{1}{\rho_{snow}h_{L,snow}L_xL_y}\left(-\max\left[q''_{inner, wood}(t),0\right](L_xL_y)-q_{cool}(t)\right)
$$
where
* $q''_{inner, wood}(t)$ is the heat flux through the wood chips at the interface wood chips/snow.
* $L_x$ and $L_y$ are the horizontal dimensions of the pile
* $h_{L,snow}$ is the latent heat of snow
* $q_{cool}(t)$ is the cooling load required to reduce the fluid temperature

This equation must be integrated in time using `solve_ivp`, requiring the definition of the problem as

$$
\frac{dH}{dt}=f(t,H)
$$

`solve_ivp` integrate this equation
$$
H(t) = \int_0^{t}\frac{dH}{dt}dt'=\int_0^tf(t',H(t'))dt'
$$

The cooling load and heat flux through the wood chips must therefore be interpolated for any time during the period of the simulation. The following code creates the function `f_cool` the cooling demand from the building as a function of time and `f_qppwood`, the heat flux (with clipped negative values) at the woodchips/snow interface as a function of time.


```python
## Reading cooling load
year = 2018
dfcool = pd.read_excel("roof_heat_rate_all.xlsx",sheet_name=str(year))
dfcool = dfcool.set_index('Unnamed: 0')
dfcool['cooling load (W)'] = dfcool['heat rate inner surface (W)'].clip(lower=0.)
t_data = np.arange(0,pd.Timedelta(dfcool.index[-1] - dfcool.index[0]).total_seconds()+1,15*60)
f_cool = interp1d(t_data,dfcool['cooling load (W)'][:].to_numpy(),kind=interp_method) #q_cool(t)
t_data_wood = np.arange(0,pd.Timedelta(dfwoodchips30cm[indexyearwood(year)].index[-1] - dfwoodchips30cm[indexyearwood(year)].index[0]).total_seconds()+1,15*60)
f_qppwood = interp1d(t_data_wood,dfwoodchips30cm[indexyearwood(year)]['melting load (W/m^2)'],kind=interp_method) #qpp_wood(t)
```

```python
def f(t,H,L_x,L_y):
    return (-f_qppwood(t)*L_x*L_y - f_cool(t))/(rho_snow*h_L_snow*L_x*L_y)

L_x = 50
L_y = 20
sol = solve_ivp(f,[0, t_data[-1]],[0],method='LSODA',t_eval=np.arange(0, t_data[-1]+1,3600),args=(L_x,L_y,))
```
